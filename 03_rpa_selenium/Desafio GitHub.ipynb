```python
"""
Projeto: Automação RPA Challenge
Descrição: Este scriptt lê uma planilha do Excel e utiliza o Selenium para
preencher automaticamente o formulário do site RPA Challenge,
pulando de linha em linha até concluir todos os cadastros
"""
from selenium import webdriver as opcoesSelenium
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
import pyautogui as tempoEspera
import openpyxl


#Abrindo o excel
caminho_arquivo = "dados.xlsx"
planilha = openpyxl.load_workbook(caminho_arquivo)
sheet = planilha.active

navegador = opcoesSelenium.Chrome()
navegador.get("https://rpachallenge.com/")
tempoEspera.sleep(3)

#Percorrer linhas pulando o cabeçalho
for linha in range(2, sheet.max_row + 1):
    
    #lendo os valordes de cada coluna
    first_name = sheet.cell(row=linha, column=1).value
    last_name = sheet.cell(row=linha, column=2).value
    company_name = sheet.cell(row=linha, column=3).value
    role = sheet.cell(row=linha, column=4).value
    address = sheet.cell(row=linha, column=5).value
    email = sheet.cell(row=linha, column=6).value
    phone = sheet.cell(row=linha, column=7).value

    print(f"Prrenchendo linha {linha}: {first_name} {last_name}")

    def preencher(campo_xpath, valor):
        elemento = navegador.find_element(By.XPATH, campo_xpath)
        elemento.clear()
        elemento.send_keys(str(valor))
        tempoEspera.sleep(1)

    preencher('//*[@ng-reflect-name="labelFirstName"]', first_name)
    preencher('//*[@ng-reflect-name="labelLastName"]', last_name)
    preencher('//*[@ng-reflect-name="labelCompanyName"]', company_name)
    preencher('//*[@ng-reflect-name="labelRole"]', role)
    preencher('//*[@ng-reflect-name="labelAddress"]', address)
    preencher('//*[@ng-reflect-name="labelEmail"]', email)
    preencher('//*[@ng-reflect-name="labelPhone"]', phone)
    
    
    #clicando em submit após preencher
    botao_submit = navegador.find_element(By.XPATH, '/html/body/app-root/div[2]/app-rpa1/div/div[2]/form/input')
    botao_submit.click()
    tempoEspera.sleep(2)
```
